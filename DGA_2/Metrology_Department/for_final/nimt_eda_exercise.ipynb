### 3.3 ทำความสะอาดชื่อเครื่องมือ DESCRIPTION / Standardize instrument names

ชื่อเครื่องมือเดียวกันอาจสะกดต่างกันหลายแบบ ต้องรวมให้เป็นชื่อเดียวกัน  
The same instrument may be spelled in different ways — we need to normalize them.

**ตัวอย่าง / Examples to normalize:**
- `LIDAR SPEED GUN`, `LIDAR GUN`, `ตรวจจับความเร็ว...LIDAR` → `'กล้องตรวจจับความเร็วชนิดมือถือ LIDAR SPEED GUN'`
- `Speed Radar (stationary)`, `Speed Radar (Stationary)` → `'Speed Radar (Stationary)'`

```python
def clean_description(desc):
    if not isinstance(desc, str):
        return 'Unknown'
    desc = desc.strip()
    if re.search(r'LIDAR SPEED\s*GUN|ตรวจจับความเร็ว.*LIDAR', desc, re.IGNORECASE) or 'LIDAR GUN' in desc.upper():
        return 'กล้องตรวจจับความเร็วชนิดมือถือ LIDAR SPEED GUN'
    if re.search(r'speed radar\s*\(station[a-z\s]*\)', desc, re.IGNORECASE):
        return 'Speed Radar (Stationary)'
    if re.search(r'Vital\s*Sign[s]?\s*Simulator', desc, re.IGNORECASE):
        return 'Vital Signs Simulator'
    return desc

df['DESCRIPTION'] = df['DESCRIPTION'].fillna('Unknown').astype(str).apply(clean_description)
print("✅ DESCRIPTION cleaned!")
```

### 6.4 Silhouette Score — วิธีที่สองในการเลือก k / Second method to choose k

Silhouette Score ยิ่งสูงยิ่งดี (ช่วง -1 ถึง 1)  
Silhouette Score: higher is better (range -1 to 1).

```python
sil_scores = []
for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    sil_scores.append(silhouette_score(X_scaled, labels))

plt.figure(figsize=(8, 4))
plt.plot(k_range, sil_scores, marker='s', color='green')
plt.title('Silhouette Score by k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.tight_layout()
plt.show()

best_k = k_range[sil_scores.index(max(sil_scores))]
print(f"Best k by Silhouette: {best_k}")
```
